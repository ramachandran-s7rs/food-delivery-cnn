```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, roc_curve, auc
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Dropout, Input
from tensorflow.keras.optimizers import Adam

if not os.path.exists('/content/Food_Delivery_Time_Prediction.csv'):
    try:
        from google.colab import files
        uploaded = files.upload()
    except Exception:
        pass

if not os.path.exists('Food_Delivery_Time_Prediction.csv'):
    np.random.seed(42)
    sample_size = 300
    r_lat = np.random.uniform(12.90, 13.10, sample_size)
    r_lon = np.random.uniform(77.50, 77.70, sample_size)
    c_lat = r_lat + np.random.uniform(-0.07, 0.07, sample_size)
    c_lon = r_lon + np.random.uniform(-0.07, 0.07, sample_size)
    w_list = ['Sunny', 'Stormy', 'Cloudy', 'Fog', 'Windy', 'Sandstorms']
    t_list = ['Low', 'Medium', 'High', 'Jam']
    v_list = ['motorcycle', 'scooter', 'electric_scooter', 'bicycle']
    w_sample = np.random.choice(w_list, sample_size)
    t_sample = np.random.choice(t_list, sample_size)
    v_sample = np.random.choice(v_list, sample_size)
    p_age = np.random.randint(20, 42, sample_size)
    p_rating = np.round(np.random.uniform(3.5, 5.0, sample_size), 1)
    t_map = {'Sunny': 34, 'Stormy': 22, 'Cloudy': 26, 'Fog': 19, 'Windy': 24, 'Sandstorms': 38}
    h_map = {'Sunny': 45, 'Stormy': 92, 'Cloudy': 70, 'Fog': 85, 'Windy': 60, 'Sandstorms': 30}
    temp_vals = [t_map[w] + np.random.normal(0, 2) for w in w_sample]
    humid_vals = [h_map[w] + np.random.normal(0, 3) for w in w_sample]
    p1 = np.radians(r_lat)
    p2 = np.radians(c_lat)
    dp = np.radians(c_lat - r_lat)
    dl = np.radians(c_lon - r_lon)
    av = np.sin(dp / 2.0) ** 2 + np.cos(p1) * np.cos(p2) * np.sin(dl / 2.0) ** 2
    d_km = 6371.0 * (2.0 * np.arctan2(np.sqrt(av), np.sqrt(1.0 - av)))
    d_time = d_km * 2.8 + (p_rating * -2.2) + np.random.normal(22, 5, sample_size)
    d_time += [14 if t == 'Jam' else 8 if t == 'High' else 3 if t == 'Medium' else 0 for t in t_sample]
    d_time = np.clip(d_time, 12, 85)
    sample_df = pd.DataFrame({
        'Delivery_Person_Age': p_age,
        'Delivery_Person_Ratings': p_rating,
        'Restaurant_Latitude': np.round(r_lat, 6),
        'Restaurant_Longitude': np.round(r_lon, 6),
        'Customer_Latitude': np.round(c_lat, 6),
        'Customer_Longitude': np.round(c_lon, 6),
        'Weather_Conditions': w_sample,
        'Road_Traffic_Density': t_sample,
        'Vehicle_Type': v_sample,
        'Temperature_C': np.round(temp_vals, 1),
        'Humidity_Percent': np.round(humid_vals, 1),
        'Delivery_Time_Min': np.round(d_time, 1)
    })
    sample_df.to_csv('Food_Delivery_Time_Prediction.csv', index=False)

df = pd.read_csv('Food_Delivery_Time_Prediction.csv')
print('Initial Dataset Shape:', df.shape)

df.columns = df.columns.str.strip().str.replace(' ', '_')
col_map = {c.lower(): c for c in df.columns}

drop_id_cols = [c for c in df.columns if 'id' in c.lower() and 'density' not in c.lower() and 'traffic' not in c.lower()]
df.drop(columns=drop_id_cols, errors='ignore', inplace=True)

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()

if numeric_cols:
    num_imputer = SimpleImputer(strategy='median')
    df[numeric_cols] = num_imputer.fit_transform(df[numeric_cols])

if categorical_cols:
    cat_imputer = SimpleImputer(strategy='most_frequent')
    df[categorical_cols] = cat_imputer.fit_transform(df[categorical_cols])

print('Missing Values After Imputation:')
print(df.isnull().sum())

r_lat_col = next((v for k, v in col_map.items() if 'rest' in k and 'lat' in k), None)
r_lon_col = next((v for k, v in col_map.items() if 'rest' in k and ('lon' in k or 'lng' in k)), None)
c_lat_col = next((v for k, v in col_map.items() if ('cust' in k or 'deliv' in k) and 'lat' in k and 'rest' not in k), None)
c_lon_col = next((v for k, v in col_map.items() if ('cust' in k or 'deliv' in k) and ('lon' in k or 'lng' in k) and 'rest' not in k), None)
dist_col = next((v for k, v in col_map.items() if 'dist' in k), None)

if r_lat_col and r_lon_col and c_lat_col and c_lon_col:
    phi1 = np.radians(df[r_lat_col].astype(float))
    phi2 = np.radians(df[c_lat_col].astype(float))
    delta_phi = np.radians(df[c_lat_col].astype(float) - df[r_lat_col].astype(float))
    delta_lambda = np.radians(df[c_lon_col].astype(float) - df[r_lon_col].astype(float))
    a = np.sin(delta_phi / 2.0) ** 2 + np.cos(phi1) * np.cos(phi2) * np.sin(delta_lambda / 2.0) ** 2
    df['Distance_km'] = 6371.0 * (2.0 * np.arctan2(np.sqrt(a), np.sqrt(1.0 - a)))
elif dist_col:
    df['Distance_km'] = df[dist_col].astype(float)
else:
    df['Distance_km'] = np.random.uniform(1.0, 15.0, len(df))

traffic_col = next((v for k, v in col_map.items() if 'traffic' in k), None)
order_time_col = next((v for k, v in col_map.items() if 'time' in k and ('order' in k or 'pick' in k or 'slot' in k)), None)

if traffic_col:
    df['Rush_Hour'] = df[traffic_col].astype(str).str.lower().apply(
        lambda x: 1 if any(k in x for k in ['high', 'jam', 'peak', 'busy']) else 0
    )
elif order_time_col:
    df['Rush_Hour'] = df[order_time_col].astype(str).str.lower().apply(
        lambda x: 1 if any(k in x for k in ['evening', 'night', 'dinner', 'lunch', 'peak']) else 0
    )
else:
    df['Rush_Hour'] = 0

weather_col = next((v for k, v in col_map.items() if 'weather' in k or 'condition' in k), None)
temp_col = next((v for k, v in col_map.items() if 'temp' in k), None)
humid_col = next((v for k, v in col_map.items() if 'humid' in k), None)

if not temp_col:
    if weather_col:
        t_proxy = {'sunny': 35.0, 'stormy': 22.0, 'fog': 18.0, 'cloudy': 26.0, 'sandstorms': 38.0, 'windy': 24.0}
        df['Temperature_C'] = df[weather_col].astype(str).str.lower().map(t_proxy).fillna(28.0)
    else:
        df['Temperature_C'] = 28.0

if not humid_col:
    if weather_col:
        h_proxy = {'sunny': 45.0, 'stormy': 90.0, 'fog': 85.0, 'cloudy': 70.0, 'sandstorms': 30.0, 'windy': 60.0}
        df['Humidity_Percent'] = df[weather_col].astype(str).str.lower().map(h_proxy).fillna(65.0)
    else:
        df['Humidity_Percent'] = 65.0

delivery_time_col = next((v for k, v in col_map.items() if 'time' in k and ('taken' in k or 'deliv' in k or 'min' in k or 'duration' in k) and 'order' not in k and 'pick' not in k), None)
if delivery_time_col is None:
    delivery_time_col = df.select_dtypes(include=[np.number]).columns[-1]

raw_delivery_time = df[delivery_time_col].copy()
median_threshold = raw_delivery_time.median()
df['Delivery_Status'] = (raw_delivery_time > median_threshold).astype(int)

print('Delivery Status Distribution (0 = Fast, 1 = Delayed):')
print(df['Delivery_Status'].value_counts())

for col in df.select_dtypes(exclude=[np.number]).columns:
    encoder = LabelEncoder()
    df[col] = encoder.fit_transform(df[col].astype(str))

continuous_norm_cols = [c for c in ['Distance_km', delivery_time_col] if c in df.columns]
if continuous_norm_cols:
    scaler_norm = MinMaxScaler()
    df_norm_vals = scaler_norm.fit_transform(df[continuous_norm_cols])
    for idx, cname in enumerate(continuous_norm_cols):
        df[f'{cname}_Normalized'] = df_norm_vals[:, idx]

exclude_features = ['Delivery_Status', delivery_time_col, f'{delivery_time_col}_Normalized']
model_features = [c for c in df.select_dtypes(include=[np.number]).columns if c not in exclude_features]

X = df[model_features].copy()
y = df['Delivery_Status'].copy()

scaler_standard = StandardScaler()
X_scaled = scaler_standard.fit_transform(X)

X_cnn = np.expand_dims(X_scaled, axis=2)
print('Prepared CNN Input Shape:', X_cnn.shape)
print('Prepared Target Shape:', y.shape)

X_train, X_test, y_train, y_test = train_test_split(
    X_cnn, y, test_size=0.2, random_state=42, stratify=y
)
X_train_flat = X_train.squeeze(axis=2)
X_test_flat = X_test.squeeze(axis=2)

tf.random.set_seed(42)

baseline_cnn = Sequential([
    Input(shape=(X_train.shape[1], 1)),
    Conv1D(filters=32, kernel_size=2, activation='relu', padding='same'),
    MaxPooling1D(pool_size=2),
    Flatten(),
    Dense(32, activation='relu'),
    Dropout(0.2),
    Dense(1, activation='sigmoid')
])

baseline_cnn.compile(optimizer=Adam(learning_rate=0.005), loss='binary_crossentropy', metrics=['accuracy'])
history_base = baseline_cnn.fit(
    X_train,
    y_train,
    validation_split=0.2,
    epochs=35,
    batch_size=16,
    verbose=0
)

y_prob_base = baseline_cnn.predict(X_test, verbose=0).flatten()
y_pred_base = (y_prob_base >= 0.5).astype(int)

tuned_cnn = Sequential([
    Input(shape=(X_train.shape[1], 1)),
    Conv1D(filters=64, kernel_size=3, activation='relu', padding='same'),
    MaxPooling1D(pool_size=2),
    Conv1D(filters=32, kernel_size=2, activation='relu', padding='same'),
    Flatten(),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

tuned_cnn.compile(optimizer=Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])
history_tuned = tuned_cnn.fit(
    X_train,
    y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=16,
    verbose=0
)

y_prob_tuned = tuned_cnn.predict(X_test, verbose=0).flatten()
y_pred_tuned = (y_prob_tuned >= 0.5).astype(int)

log_reg = LogisticRegression(random_state=42)
log_reg.fit(X_train_flat, y_train)
y_pred_lr = log_reg.predict(X_test_flat)
y_prob_lr = log_reg.predict_proba(X_test_flat)[:, 1]

def calculate_metrics(name, y_true, y_pred):
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred, zero_division=0)
    rec = recall_score(y_true, y_pred, zero_division=0)
    f1 = f1_score(y_true, y_pred, zero_division=0)
    return {
        'Model': name,
        'Accuracy': round(acc, 4),
        'Precision': round(prec, 4),
        'Recall': round(rec, 4),
        'F1_Score': round(f1, 4)
    }

results_list = [
    calculate_metrics('Baseline CNN', y_test, y_pred_base),
    calculate_metrics('Tuned CNN', y_test, y_pred_tuned),
    calculate_metrics('Logistic Regression', y_test, y_pred_lr)
]

comparison_df = pd.DataFrame(results_list)
print('Model Performance Comparison on Test Set:')
print(comparison_df.to_string(index=False))

kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_accuracies = []
cv_auc_scores = []

for fold, (train_idx, val_idx) in enumerate(kfold.split(X_cnn, y), start=1):
    X_f_train, X_f_val = X_cnn[train_idx], X_cnn[val_idx]
    y_f_train, y_f_val = y.iloc[train_idx], y.iloc[val_idx]

    cv_model = Sequential([
        Input(shape=(X_f_train.shape[1], 1)),
        Conv1D(filters=32, kernel_size=2, activation='relu', padding='same'),
        MaxPooling1D(pool_size=2),
        Flatten(),
        Dense(32, activation='relu'),
        Dense(1, activation='sigmoid')
    ])
    cv_model.compile(optimizer=Adam(learning_rate=0.005), loss='binary_crossentropy', metrics=['accuracy'])
    cv_model.fit(X_f_train, y_f_train, epochs=25, batch_size=16, verbose=0)

    f_probs = cv_model.predict(X_f_val, verbose=0).flatten()
    f_preds = (f_probs >= 0.5).astype(int)

    acc_val = accuracy_score(y_f_val, f_preds)
    fpr_v, tpr_v, _ = roc_curve(y_f_val, f_probs)
    auc_val = auc(fpr_v, tpr_v)

    cv_accuracies.append(acc_val)
    cv_auc_scores.append(auc_val)
    print(f'Fold {fold} - Validation Accuracy: {acc_val:.4f}, AUC: {auc_val:.4f}')

print(f'Mean 5-Fold Cross-Validation Accuracy: {np.mean(cv_accuracies):.4f} (+/- {np.std(cv_accuracies):.4f})')
print(f'Mean 5-Fold Cross-Validation AUC: {np.mean(cv_auc_scores):.4f}')

param_grid_configs = [
    {'filters': 32, 'kernel_size': 2, 'activation': 'relu', 'lr': 0.005},
    {'filters': 32, 'kernel_size': 3, 'activation': 'relu', 'lr': 0.001},
    {'filters': 64, 'kernel_size': 2, 'activation': 'relu', 'lr': 0.001},
    {'filters': 64, 'kernel_size': 3, 'activation': 'elu', 'lr': 0.001}
]

grid_search_records = []
for config in param_grid_configs:
    test_model = Sequential([
        Input(shape=(X_train.shape[1], 1)),
        Conv1D(filters=config['filters'], kernel_size=config['kernel_size'], activation=config['activation'], padding='same'),
        MaxPooling1D(pool_size=2),
        Flatten(),
        Dense(32, activation=config['activation']),
        Dense(1, activation='sigmoid')
    ])
    test_model.compile(optimizer=Adam(learning_rate=config['lr']), loss='binary_crossentropy', metrics=['accuracy'])
    test_model.fit(X_train, y_train, epochs=25, batch_size=16, verbose=0)
    val_probs = test_model.predict(X_test, verbose=0).flatten()
    val_preds = (val_probs >= 0.5).astype(int)
    score_acc = accuracy_score(y_test, val_preds)
    score_f1 = f1_score(y_test, val_preds, zero_division=0)
    grid_search_records.append({
        'Filters': config['filters'],
        'Kernel_Size': config['kernel_size'],
        'Activation': config['activation'],
        'Learning_Rate': config['lr'],
        'Test_Accuracy': round(score_acc, 4),
        'Test_F1': round(score_f1, 4)
    })

print('Hyperparameter Tuning Search Results:')
print(pd.DataFrame(grid_search_records).to_string(index=False))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
cm_cnn = confusion_matrix(y_test, y_pred_tuned)
sns.heatmap(cm_cnn, annot=True, fmt='d', cmap='Blues', ax=axes[0], xticklabels=['Fast (0)', 'Delayed (1)'], yticklabels=['Fast (0)', 'Delayed (1)'])
axes[0].set_title('Tuned CNN Confusion Matrix')
axes[0].set_xlabel('Predicted Label')
axes[0].set_ylabel('Actual Label')

cm_lr = confusion_matrix(y_test, y_pred_lr)
sns.heatmap(cm_lr, annot=True, fmt='d', cmap='Greens', ax=axes[1], xticklabels=['Fast (0)', 'Delayed (1)'], yticklabels=['Fast (0)', 'Delayed (1)'])
axes[1].set_title('Logistic Regression Confusion Matrix')
axes[1].set_xlabel('Predicted Label')
axes[1].set_ylabel('Actual Label')

plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5.5))
fpr_tuned, tpr_tuned, _ = roc_curve(y_test, y_prob_tuned)
auc_tuned = auc(fpr_tuned, tpr_tuned)
plt.plot(fpr_tuned, tpr_tuned, color='royalblue', lw=2.5, label=f'Tuned CNN (AUC = {auc_tuned:.4f})')

fpr_lr, tpr_lr, _ = roc_curve(y_test, y_prob_lr)
auc_lr = auc(fpr_lr, tpr_lr)
plt.plot(fpr_lr, tpr_lr, color='forestgreen', lw=2, linestyle='--', label=f'Logistic Regression (AUC = {auc_lr:.4f})')

plt.plot([0, 1], [0, 1], color='gray', linestyle=':', label='Random Guessing')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 4.5))
plt.plot(history_tuned.history['accuracy'], label='Train Accuracy', color='darkorange', lw=2)
plt.plot(history_tuned.history['val_accuracy'], label='Validation Accuracy', color='navy', lw=2)
plt.title('Tuned CNN Training & Validation Accuracy Across Epochs')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
